## Main concepts in Reinforcement Learning

Reinforcement learning (RL) is a type of machine learning that does not employ a predefined dataset, unlike supervised and unsupervised learning. RL involves two fundamental components: the *Agent* and the *Environment*. 

The agent is the entity that learns, while the environment is the domain where the agent operates. You can think of the agent as the character in a game, while the environment is the game itself. The agent interacts with the environment through *actions*. Agents pick actions using a *policy*. A simple, although probably not very successful, policy could be to just always go left. 

When an action is performed on the environment, the environment returns a new *state* (how the environment looks after the action) and a *reward*. Rewards are what allows the agent to learn how to act. Positive rewards encourage similar behavior, while negative rewards discourage behaviour. This experience-based learning method makes RL particularily effective for tasks like teaching robots how to walk or teaching machines how to play games.

### Exercise 2: Policy evaluation - Deterministic policies

Now that you are familiar with environment setup and agent movement, let's look at how to automate it! RL would not be very practical if a human had to dictate each move. 

Agents move by using *policies*: Given an environment state, the policy will return an action for the agent to perform. Good policies are crucial for the agent's success, but determining how good a policy is in a given environment is not trivial. The act of determining a policy's *goodness* is called policy evaluation. We use policy evaluation to select the most successful policy for a given problem or task.

Policies can be *deterministic* or *stochastic*. A deterministic policy will always chose the same action in the same situation (state). A stochastic policy introduces variability and can choose different actions even in the same situation. We will evaluate deterministic policies by computing the overall return of the policy, while we will use the average overall return for stochastic policies.

#### Exercise 2d: Policy return

At the beginning of this section, we talked about calculating the *return* of policies. The return of a policy differs from the *reward* of an action. Rewards quantify the short-term (local) *goodness* of a single action, while the return quantify the long-term (global) goodness of a policy. 

The return of a deterministic policy is calculated based on both immediate and expected future rewards. A discount factor $\gamma<1$ is used to make immediate rewards count more than future rewards. One reason for this is that rewards we expect to get in the future are more uncertain than rewards closer in time.

The overall return of a policy $\pi$ from time $t$ using a discount $\gamma$ can be given as:
$$
G_{\pi}(t) = \sum_{t} \gamma^t r(t),
$$
where $G_{\pi}(t)$ is the return at time $t$ and $r(t)$ is the reward at time $t$ under policy ${\pi}$. 

Use the trajectory data from your last run to compute the return of policy `plcs.policy1()` from the initial state at $t=0$ with a discount rate of $\gamma = 0.9$.

#### Exercise 3b: Monte Carlo Policy Evaluation

As for deterministic policies, we need to be able to quantify the *goodness* of a stochastic policy. As each simulation of a stochastic policy might return different results due to the randomness-factor, the formula for computing the return of a deterministic policy will not work for stochastic policies (each simulation would yield a different return value). An intuitive solution to this problem would be to compute the *expected* return of the stochastic policy.

To do this, we employ *Monthe Carlo Policy Evaluation*. Monte Carlo methods constitute a wide family of techniques based on simulation and estimation of expected quantities. These methods are useful for systems that are hard to analyze or have unknown dynamics, but are easy or cheap to simulate. By collecting samples from repeated simulations, we can compute statistical quantities about the system.

As the dynamics of our enviroment (how we transition from one state to another given an action) are known, we could theoretically comute the analytical value of the policy, but we will assume this computation is too expensive and instead rely on Monte Carlo approximation.

To compute the expected return of a stochastic policy, we need to collect trajectories from multiple episodes. We use a set of trajectories to compute expected return of the policy.

Monte Carlo is a tabular approach to policy evaluation. There are of course other possible approaches to policy evaluation of stochastic policies, for instance *dynamic programming* and *temporal differences*. We will not get into these methods in this section, but *Q-Learning*, which is a temporal difference reinforcement learning algorithm, is covered in a later exercise.

### Exercise 4: Policy improvement - Monte Carlo Policy Evaluation for State Values

We now know how good a given policy is, but this knowledge is not that helpful if we don't know how to improve said policy. Previously, we have only looked at the overall or *global* goodness value. Finding a good, or even optimal, policy by creating a lot of different policies and then comparing their return is tedious, ineffective, and, in a lot of cases, impossible. The number of possible policies drastically increas with the number of states and actions.

To improve a policy, we instead need to look at *local* goodness values. To do this, we use *state values* $v(s)$, which evaluate how good a state is (expected return from a given state $s$), or *action-state values* $q(s,a)$, evaluating how good an action $a$ is in a given state $s$ (expected return from state $s$ by performing action $a$). By knowing the goodness of a state or action, we can guide the agent towards better choices.

In this section, we will look at state values. The state value $v(s)$ of a state $s$ under the policy $\pi$ is defined as the return starting from state $s$ (as opposed to the return starting from the initial state $s_0$).
$$
v_{\pi}(s) = \mathbb{E} \left[ G(t) \vert S(t)=s \right],
$$

We will use Monte Carlo to compute the return, as we did for the overall return. We compute state values by running multiple episodes in the environment, using a completely random policy. We start with a completely random policy to mirror our lack of knoledge about the enviroment and allow the agent to explore the environment widely.

### Exercise 5: Q-Learning

In our last task, we looked at Monte Carlo Policy Evaluation for State Values, briefly mentioning that other approaches exist. *Q-Learning* is an *off-policy*, *temporal difference* reinforcement learning algorithm that works by estimating *state-acton values* $q(s,a)$. Off-policy means that the agent behaves according to a behaviour policy $\pi_b$, while learning an optimal target policy $\pi_t$, while temporal difference means that the agent uses an algorithm that allows learning in real-time withougt needing to collect whole episodes. In the case of Monte Carlo, we saw that the agent only used one policy and we had to first collect a set of episodes, before using these episodes to create an improved policy. A sub-optimal policy is usually better for exploration and learning than an approximation of the optimal policy in the environment.

Implement and run Q-learning for one episode:
1. Given that we are in state $s$, we choose an action $a$ according to our behaviour policy *behaviour_policy()*
2. We take action $a$, receive reward $r$ and end up in state $s'$
3. We update the action-value function $q(s,a)$ according to the following formula:
$$
q(s,a) \leftarrow q(s,a) + \alpha \left[ r + \gamma \max_{a'} q(s',a') - q(s,a) \right]
$$
where $\max_{a'} q(s',a')$ is the action-value of the action $a'$ that maximizes the action-value for state $s'$.

So far we have solved the reinforcement learning problem using a *tabular* approach. We have used matrices to store state-values, $v$, and action-values, $q$. The tabular approach is simple and interpretable, but the state-action space grows exponentially with the number of states and actions. Large spaces are memory-intensive and time consuming to explore. Tabular methods require discrete states and actions and are not suitable for continous tasks. Tabular methods also lack generalization. They treat every state independently, and are not able to benefit from previous exploration in similar states. This differs from for instance deep reinforcement learning which employs neural networks to generalize (function approximation).
