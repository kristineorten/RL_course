## Main concepts in Reinforcement Learning

Reinforcement learning (RL) is a type of machine learning that does not employ a predefined dataset, unlike supervised and unsupervised learning. RL involves two fundamental components: the *Agent* and the *Environment*. 

The agent is the entity that learns, while the environment is the domain where the agent operates. You can think of the agent as the character in a game, while the environment is the game itself. The agent interacts with the environment through *actions*. Agents pick actions using a *policy*. A simple, although probably not very successful, policy could be to just always go left. 

When an action is performed on the environment, the environemnt returns the new *state* of the environment (how the environment looks after the action) and a *reward*. Rewards are whan allows the agent to learn how to act. Positive rewards encourage similar behavior, while negative rewards discourage behaviour. This experience-based learning method makes RL particularily effective for tasks like teaching robots how to walk or teaching machines how to play games.

### Exercise 1b: Policy evaluation - Deterministic policies
Now that you are familiar with environment setup and agent movement, let's look at how to automate it! RL would not be very practical if a human had to dictate each move. 

Agents move by using *policies*: Given an environment state, the policy will return an action for the agent to perform. Good policies are crucial for the agent's success, but determining how good a policy is in a given environment is not trivial. The act of determining a policy's *goodness* is called policy evaluation. We use policy evaluation to select the most successful policy for a given problem or task.

Policies can be *deterministic* or *stochastic*. A deterministic policy will always chose the same action in the same situation (state). A stochastic policy introduces variability and can choose different actions even in the same situation. We will evaluate deterministic policies by computing the overall return of the policy, while we will use the average overall return for stochastic policies.

At the beginning of this section, we talked about calculating the *return* of policies. The return of a policy differs from the *reward* of an action. Rewards quantify the short-term (local) *goodness* of a single action, while the return quantify the long-term (global) goodness of a policy. We use a formula for this:

Overall return of a policy $\pi$ from time $t$ using a discount $\gamma$ can be given as:
$$
G_{\pi}(t) = \sum_{t} \gamma^t r(t),
$$
where $G_{\pi}(t)$ is the return at time $t$ and $r(t)$ is the reward at time $t$ under policy ${\pi}$. 

We will not talk too much about what this formula means, but in short the return of a deterministic policy is calculated based on both immediate and expected future rewards. The discount factor $\gamma<1$ is used to make immediate rewards count more than future rewards. One reason for this is that rewards we expect to get in the future are more uncertain than rewards closer in time.

As for deterministic policies, we need to be able to quantify the *goodness* of a stochastic policy. As each simulation of a stochastic policy might return different results due to the randomness-factor, the formula for computing the return of a deterministic policy will not work for stochastic policies (each simulation would yield a different return value). An intuitive solution to this problem would be to compute the *expected* return of the stochastic policy.

To do this, we employ *Monthe Carlo Policy Evaluation*. **Monte Carlo methods** constitute a wide family of techniques based on simulation and estimation of expected quantities. These methods are useful for systems that are hard to analyze or have unknown dynamics, but are easy or cheap to simulate. By collecting samples from repeated simulations, we can compute statistical quantities about the system.

As the dynamics of our enviroment (how we transition from one state to another given an action) are known, we could theoretically comute the analytical value of the policy, but we will assume this computation is too expensice and instead rely on Monte Carlo approximation.
